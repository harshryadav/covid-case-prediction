# GluonTS DeepAR API Demonstration

This notebook demonstrates the **DeepAR model API** in GluonTS.

**Purpose**: Minimal demonstration of the DeepAR estimator interface and our wrapper functions.

**Scope**: 
- Uses toy/synthetic data for quick execution
- Shows API configuration options
- Demonstrates basic training and forecasting

**For complete COVID-19 example**: See `GluonTS_DeepAR.example.ipynb`

## 1. Setup and Imports

```python
import sys
sys.path.append('.')

import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Import our wrapper functions
from utils.gluonts_utils import (
    create_gluonts_dataset,
    train_deepar,
    generate_forecast,
    evaluate_forecast,
    plot_forecast
)

print("✓ Imports successful")
```

## 2. Create Toy Dataset

Generate synthetic time series data for demonstration.

```python
# Generate 100 days of synthetic data
np.random.seed(42)
dates = pd.date_range('2020-01-01', periods=100, freq='D')

# Create trend + seasonality + noise
trend = np.linspace(100, 1000, 100)
seasonality = 200 * np.sin(np.linspace(0, 4*np.pi, 100))
noise = np.random.normal(0, 50, 100)
values = trend + seasonality + noise
values = np.maximum(values, 0)  # Keep non-negative

# Create DataFrame
toy_df = pd.DataFrame({
    'Date': dates,
    'value': values
})

print(f"✓ Created toy dataset: {len(toy_df)} days")
print(f"  Value range: {toy_df['value'].min():.0f} to {toy_df['value'].max():.0f}")
toy_df.head()
```

## 3. Prepare GluonTS Dataset

Convert pandas DataFrame to GluonTS `ListDataset` format.

```python
# Split into train/test
train_df = toy_df.iloc[:80]
test_df = toy_df.iloc[60:]

# Create GluonTS datasets
train_ds = create_gluonts_dataset(
    df=train_df,
    target_column='value',
    freq='D',
    prediction_length=14
)

test_ds = create_gluonts_dataset(
    df=test_df,
    target_column='value',
    freq='D',
    prediction_length=14
)

print("✓ GluonTS datasets created")
print(f"  Train: {len(train_df)} days")
print(f"  Test: {len(test_df)} days")
```

## 4. Train DeepAR Model

Demonstrate the `train_deepar()` API with key configuration options.

```python
# Train with minimal configuration (fast for demo)
predictor = train_deepar(
    train_data=train_ds,
    prediction_length=14,      # Forecast 14 days
    context_length=30,         # Use 30 days of history
    num_layers=1,              # Single RNN layer
    num_cells=20,              # 20 hidden units
    epochs=5,                  # Quick training
    batch_size=32,
    learning_rate=0.001
)

print("\n✓ DeepAR model trained successfully!")
```

## 5. Generate Forecasts

Use the trained model to generate probabilistic forecasts.

```python
# Generate forecasts
forecasts, ground_truths = generate_forecast(
    predictor=predictor,
    test_data=test_ds,
    num_samples=100  # Monte Carlo samples for uncertainty
)

forecast = forecasts[0]
ground_truth = ground_truths[0]

print("\n✓ Forecasts generated")
print(f"  Forecast length: {len(forecast.mean)} days")
print(f"  Mean prediction (first 5): {forecast.mean[:5]}")
print(f"  10th percentile (first 5): {forecast.quantile(0.1)[:5]}")
print(f"  90th percentile (first 5): {forecast.quantile(0.9)[:5]}")
```

## 6. Evaluate Performance

Compute forecast accuracy metrics.

```python
# Evaluate forecast
metrics = evaluate_forecast(
    forecast=forecast,
    ground_truth=ground_truth
)

print("\nModel Performance:")
print("=" * 50)
print(f"MAE:  {metrics['mae']:.2f}")
print(f"RMSE: {metrics['rmse']:.2f}")
print(f"MAPE: {metrics['mape']:.2f}%")
print(f"CRPS: {metrics['crps']:.2f}")
print("=" * 50)
```

## 7. Visualize Forecast

Plot forecast with confidence intervals.

```python
# Plot forecast
plot_forecast(
    forecast=forecast,
    ground_truth=ground_truth,
    title="DeepAR Toy Example Forecast",
    save_path="deepar_api_demo.png"
)

print("\n✓ Forecast plotted and saved to 'deepar_api_demo.png'")
```

## 8. Summary

### DeepAR API Highlights

**Key Configuration Parameters**:
- `prediction_length`: How many steps to forecast
- `context_length`: Historical data to use as input
- `num_layers`: Number of RNN layers (1-3 typical)
- `num_cells`/`hidden_size`: Hidden layer size (20-100 typical)
- `epochs`: Training iterations (5-50 typical)
- `learning_rate`: Optimizer learning rate (0.001 typical)

**What DeepAR Provides**:
- Probabilistic forecasts (not just point estimates)
- Confidence intervals (10-90%, 25-75%)
- Handles multiple time series
- Captures temporal dependencies via RNN

**Best For**:
- Sequential time series with dependencies
- When uncertainty quantification is important
- Medium to complex temporal patterns
